Gene selection criterion:

_log2FoldChange: the direction and intensity on a log2 scale: +2 means 4x more expression (here, in hypoxia), -3 means 8x less. It is an estimate with uncertainty that is captured by the lfcSE column (standard error). 

Example: a gene with log2FC = 4.9 but lfcSE = 1.3 is less trustworthy than log2FC = 2.7 and lfcSE = 0.26

_padj: probability of observing counts this extreme assuming the gene is not DE

baseMean: the mean of the normalized counts for that gene, averaged accross all samples, both conditions (O2 and no O2 pooled). Tells us how much the gene is expressed overall, i.e how much data underlies the statistics. 

    Low baseMean : the fold change is computed from a handfull of reads, so the estimate is noisy. 

    High baseMean: the gene is abundantly expressed and the fold change is solid. Also an abundant transcript should be easier to validate by RT-qPCR or Northern blot

```python
hypoxia_data = pd.read_csv("rnaseq_hypoxia.csv", sep=";", index_col=0)
normoxia_data = pd.read_csv("rnaseq_normoxia.csv", sep=";", index_col=0)
overview_data = pd.read_csv("01_complete_gene_ranking.csv", sep=",", index_col=0)
```

```python
#filter choices
basemean = 200
fold_change_hypoxia = 2.5
fold_change_normoxia = -2.5
p_adj =  1e-6
```

```python
normoxia_query = "baseMean >= @basemean & log2FoldChange <= @fold_change_normoxia & padj <= @p_adj"
hypoxia_query = "baseMean >= @basemean & log2FoldChange >= @fold_change_hypoxia & padj <= @p_adj"
```

```python
print(len(overview_data.query(hypoxia_query)))
print(overview_data.query(hypoxia_query))
```
